```python
import pandas as pd
import os
import numpy as np
```

```python


data_path = "../data"

customers = pd.read_csv(f"{data_path}/customers.csv")
products = pd.read_csv(f"{data_path}/products.csv")
orders = pd.read_csv(f"{data_path}/orders.csv")
order_items = pd.read_csv(f"{data_path}/order_items.csv")
payments = pd.read_csv(f"{data_path}/payments.csv")
returns = pd.read_csv(f"{data_path}/returns.csv")
shipping = pd.read_csv(f"{data_path}/shipping.csv")
discounts = pd.read_csv(f"{data_path}/discounts.csv")
regions = pd.read_csv(f"{data_path}/regions.csv")
```

```python
datasets["order_items"]["discount_pct"] = datasets["order_items"]["discount_pct"].fillna(datasets["order_items"]["discount_pct"].mean())
```

```python
datasets["order_items"]["gross_profit"] = (
    datasets["order_items"]["net_sales"]
    - datasets["order_items"]["estimated_cost"]
)
```

```python
datasets["order_items"]["profit_margin_pct"] = np.where(order_items["net_sales"] > 0,
    datasets["order_items"]["gross_profit"]
    / datasets["order_items"]["net_sales"] * 100,
    np.nan
)
```

```python
order_metrics = (
    datasets["order_items"]
    .groupby("order_id")
    .agg(
        gross_sales=("unit_price", lambda x: 0),
        net_sales=("net_sales", "sum"),
        estimated_cost=("estimated_cost", "sum"),
        gross_profit=("gross_profit", "sum")
    )
    .reset_index()
)
```

```python
order_metrics.sample(4)
```

```python
datasets["order_items"]["gross_sales"] = (datasets["order_items"]["quantity"]* datasets["order_items"]["unit_price"])
```

```python
order_metrics = (
    datasets["order_items"]
    .groupby("order_id")
    .agg(
        gross_sales=("gross_sales", "sum"),
        net_sales=("net_sales", "sum"),
        estimated_cost=("estimated_cost", "sum"),
        gross_profit=("gross_profit", "sum")
    )
    .reset_index()
)
```

```python
order_metrics["profit_margin_pct"] = np.where(
    order_metrics["net_sales"] > 0,
    order_metrics["gross_profit"]
    / order_metrics["net_sales"] * 100,
    np.nan
)
```

```python
orders = orders.merge(
    order_metrics,
    on="order_id",
    how="left"
)
```

```python
datasets["shipping"]["carrier"] = datasets["shipping"]["carrier"].fillna("Unknown")
```

```python
datasets["shipping"]["delivery_delay_days"] = (
    datasets["shipping"]["actual_delivery_date"]
    - datasets["shipping"]["promised_delivery_date"]
).dt.days
```
